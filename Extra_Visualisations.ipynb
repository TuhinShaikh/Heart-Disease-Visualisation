```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/heart_clean.csv")  # load cleaned data
```

```python
plt.figure(figsize=(6, 6)) #checking data
counts = df.groupby("HeartDiseaseLabel").size()
plt.pie(counts.values, labels=counts.index, autopct="%1.1f%%", startangle=90, colors=["tab:orange", "tab:blue"])
plt.title("Share of Patients With and Without Heart Disease")
```

```python
plt.figure(figsize=(6, 6)) #Gender Split
sx = df.groupby("Sex").size()
plt.pie(sx.values, labels=["Female", "Male"], autopct="%1.1f%%", startangle=90, wedgeprops={"width": 0.4})
plt.title("Gender Split of Patients in the Dataset")
```

```python
plt.figure(figsize=(9, 7)) #Correlation heatmap
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap of Numeric Columns")
```

```python
plt.figure(figsize=(9, 6)) #Age vs max heart rate scatter
sns.scatterplot(data=df, x="Age", y="MaxHR", hue="HeartDiseaseLabel", alpha=0.7)
plt.title("Age vs Maximum Heart Rate by Heart Disease Status")
plt.xlabel("Age (years)")
plt.ylabel("Maximum Heart Rate (bpm)")
```

```python
sns.lmplot(data=df, x="Age", y="MaxHR", hue="HeartDiseaseLabel", height=6, aspect=1.4)
plt.title("Age vs Maximum Heart Rate with Trend Lines")
plt.xlabel("Age (years)")
plt.ylabel("Maximum Heart Rate (bpm)")
```

```python
plt.figure(figsize=(7, 5))  #Patients by gender and disease
sns.countplot(data=df, x="Sex", hue="HeartDiseaseLabel")
plt.title("Number of Patients by Gender and Heart Disease Status")
plt.xlabel("Gender (F = Female, M = Male)")
plt.ylabel("Number of Patients")
```

```python
plt.figure(figsize=(8, 5))  # age difference
sns.violinplot(data=df, x="HeartDiseaseLabel", y="Age")
plt.title("Age Distribution by Heart Disease Status")
plt.xlabel("Heart Disease Status")
plt.ylabel("Age (years)")
```
